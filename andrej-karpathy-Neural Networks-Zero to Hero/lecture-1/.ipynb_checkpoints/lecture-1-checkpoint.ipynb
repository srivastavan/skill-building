## Let's first understand what a derivative is and exactly what information it gives you.

Let's define a scalar value function f(x), which takes a single scalar x and returns a single scalar y.

```python
def f(x):
    return 3*x**2 - 4*x + 5
```

```python
f(3.0)
```

We can also plot this function. We can understand probably from its shape x**2 that is a parabola (it's a quadratic function).

```python
xs = np.arange(-5, 5, 0.25)
ys = f(xs)

plt.plot(xs, ys)
plt.show()
```

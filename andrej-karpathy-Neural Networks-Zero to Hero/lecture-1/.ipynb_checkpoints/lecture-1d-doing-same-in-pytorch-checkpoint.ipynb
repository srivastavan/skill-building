# Implementing the Value class and backward logic using Pytorch

```python
## Import the library
import torch
```

```python
## To create variable, we create tensors in Pytorch

# If I just replicate the same logic using Tensor, then we will get this result.

## inputs (2-D nuerons)
# x1 = Value(2.0, label = "x1") --- Value class
x1 = torch.tensor(2.0)
x1.requires_grad = True
x2 = torch.tensor(0.0)
x2.requires_grad = True

w1 = torch.tensor(-3.0)
w1.requires_grad = True

w2 = torch.tensor(1.0)
w2.requires_grad = True

b = torch.tensor(6.8813735870195432)
b.requires_grad = True

n = x1*w1 + x2*w2 + b
o = torch.tanh(n)

o.backward()

print('---')
print('x1', x1.grad.item())
print('x2', x2.grad.item())
print('w1', w1.grad.item())
print('w2', w2.grad.item())
```

### the documentation also shows, creating tensor like torch.tensor(2) creates a zero-dimensional (scalar) tensor. This will cause errors later when I want to implement matrix multiplication.

```python
# Let's fix this.

x1 = torch.tensor([2.0], dtype = torch.double, requires_grad = True)
# by default it uses single-precision fload `float32`, so we are casting it here to use double precision float `float64`
```

```python
x1.shape
```

```python
# let's write the full logic

x1 = torch.tensor([2.0], dtype = torch.double, requires_grad = True)
x2 = torch.tensor([0.0], dtype = torch.double, requires_grad = True)
w1 = torch.tensor([-3.0], dtype = torch.double, requires_grad = True)
w2 = torch.tensor([1.0], dtype = torch.double, requires_grad = True)
b = torch.tensor([6.8813735870195432], dtype = torch.double, requires_grad = True)

n = x1*w1 + x2*w2 + b
o = torch.tanh(n)

print(o.data) # this is a tensor
print(o.data.item()) # prints the actual number
o.backward()

print('---')
print('x1', x1.grad.item())
print('x2', x2.grad.item())
print('w1', w1.grad.item())
print('w2', w2.grad.item())
```

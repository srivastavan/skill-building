## Let's first understand what a derivative is and exactly what information it gives you.

Let's define a scalar value function f(x), which takes a single scalar x and returns a single scalar y.

```python
def f(x):
    return 3*x**2 - 4*x + 5
```

```python
f(3.0)
```

We can also plot this function. We can understand probably from its shape x**2 that is a parabola (it's a quadratic function).

```python
xs = np.arange(-5, 5, 0.25)
ys = f(xs)

plt.plot(xs, ys)
plt.show()
```

## What is derivative?

Let's look at the definition.

- Definition: A derivative measures how much a function's output changes when its input changes a tiny amount.
- Geometric meaning: It represents the slope of the tangent line to a curve at a specific point.
- Formula: Calculated as the limit of the difference quotient as the interval h approaches zero:

  $$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

```python
# let's say h = 0.001
h = 0.001
x = 3.0
f(x)
```

Looking at the plot, let's see how much do you expect the function f(x+h)?

Do you expect it to be slightly greater than 20, or less than 20? 

Looking at the plot, if we slightly go positively, the function will respond positively, so it will be greater than 20.

```python
f(x+h)
```

Okay, so to find how much the function responded, we can find the difference how see the strength of the slope or the size of the slope.

```python
f(x+h) - f(x)
```

The number is positive, so that means we are moving upward. Next, we need to normalize this by the run, so we have the rise over run to get the slope (so let's divide by h).

```python
d = (f(x+h)-f(x))/h
```

```python
d
```

This value is a numerical approximation of the slope, because we have to make h very very small to converge to the exact amount.

```python
h = 0.00001
x = 3.0
(f(x+h)-f(x))/h
```

## Complex function

The complex function d is a function of three scalar variables. Now, we would like to look at the derivative of d with respect to a, b and c.


### derivative with respect to a

```python
h = 0.00001

# inputs
a = 2.0
b = -3.0
c = 10.0

d1 = a*b + c
a += h

d2 = a*b + c

print('d1', d1)
print('d2', d2)
# print('slope', (d2-d1)/h)
```

Let's try to understand this here. When a will be bumped by h and all the input remain constant, so it will be 

$$ d2 = (2+h) * -3 + 10$$

Okay, when a increases by **a** small tiny amount *h*, then the product of (a+h)*b also increases, but because b is negative, this value is negative, and therefore, d2 will be less than d1, and the slope will be negative. 

```python
print('slope', (d2-d1)/h)
```

### derivative with respect to b

```python
h = 0.00001

# inputs
a = 2.0
b = -3.0
c = 10.0

d1 = a*b + c
b += h

d3 = a*b + c
```

Let's try to understand this here. When b will be bumped by h and all the input remain constant, so it will be 

$$ d2 = (a) * (-3+h) + 10$$

So, (b+h) will be a less value, and therefore the product (a*(b+h)) will be small, since this is negative, and overall d3 will be positive, and the slope will be positive.

print('d1', d1)
print('d3', d3)
print('slope', (d3-d1)/h)

### derivative with respect to c

```python
h = 0.00001

# inputs
a = 2.0
b = -3.0
c = 10.0

d1 = a*b + c
c += h

d4 = a*b + c
```

Here c is a constant, so if I add a tiny amount to c, it will raise the value by exactly the same tiny amount, as a and b are constant. So slope will be 1.

```python
print('d1', d1)
print('d4', d4)
print('slope', (d4-d1)/h)
```

```python
## Class Value takes a single scalar value that is wraps and keep tracks of
class Value:

    def __init__(self, data):
        self.data = data

    def __repr__(self):
        return f"Value(data = {self.data})" ## this is for printing the value
```

### Step 2: Now let's add more functions in our Value class, for example, we want to add a + b, a * b

```python
a + b
```

Python doesn't know how to add these value objects, so we need to tell it.

So we will use these special double underscore methods in python (\__add__) to define these operators so these objects.

```python
class Value:

    def __init__(self, data):
        self.data = data

    def __repr__(self):
        return f"Value(data = {self.data})" ## this is for printing the value

    def __add__(self, other):
        out = Value(self.data + other.data)
        return out

    def __mul__(self, other):
        out = Value(self.data * other.data)
        return out

a = Value(2.0)
b = Value(-3.0)
c = Value(10.0)
a + b   ## python will internally call a.__add__(b)
a * b + c ## python will internally call a.__mul__(b).__add__(c)

# a.__mul__(b).__add__(c)
```

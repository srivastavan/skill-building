# Summary of what we learned

## What are Neural Nets?

Neural network are these mathematical expressions that take input as the data, and they take input the wieights and the parameters of the neural net. They take the mathematical expression for the forward pass, followed by a loss function. A loss function tries to measure the accuracy of how close the predictions are to the actual inputs. We then back-propogate and calculate the gradient of every paramemter in our NN w.r.t to the loss, and then we update our weights based on the direction of these gradients using gradient descent.
